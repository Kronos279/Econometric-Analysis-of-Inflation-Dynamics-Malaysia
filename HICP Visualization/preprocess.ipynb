```python
import pandas as pd

df = pd.read_csv('CPM15.csv')

df.head(100)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.strip().str.replace(" ", "_")

text_cols = ['Statistic_Label', 'Month', 'Commodity_Group', 'UNIT']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.replace(r'\.\.\.$', '', regex=True)

df[['Year', 'Month_Name']] = df['Month'].str.split(' ', n=1, expand=True)
df['Year'] = pd.to_numeric(df['Year'], errors='coerce')

df['Date'] = pd.to_datetime(df['Month_Name'] + ' ' + df['Year'].astype(str),format='%B %Y', errors='coerce')

df['VALUE'] = pd.to_numeric(df['VALUE'], errors='coerce')

df = df.dropna(subset=['VALUE'])

df = df.drop_duplicates()

print(df.info())
print(df.isnull().sum())

df.to_csv("hicp_data_cleaned.csv", index=False)
```
